### 3. Визуализация

Используя библиотеки matplotlib/sns/plotly/pandas и набор данных представленный ранее визуализируйте

````python
import matplotlib.pyplot as plt
````

1. Как менялась средняя цена квартиры в Нагатинском Затоне(Nagatinskij Zaton) с течением времени

````python
df_nag = df[df['sub_area'] == 'Nagatinskij Zaton']
df_nag['timestamp'] = pd.to_datetime(df_nag['timestamp'])
mean_price = df_nag.groupby('timestamp')['price_doc'].mean()

plt.figure(figsize=(12, 6))

plt.plot(mean_price.index, mean_price.values)

plt.title('Динамика средней цены квартир в Нагатинском Затоне')
plt.xlabel('Дата')
plt.ylabel('Средняя стоимость, ₽')

plt.grid()
plt.show()
````

````python
# по месяцам
df_nag['timestamp'] = pd.to_datetime(df_nag['timestamp'])

mean_price = df_nag.groupby(
    pd.Grouper(key='timestamp', freq='ME')
)['price_doc'].mean().dropna()

plt.figure(figsize=(16, 7))

plt.plot(mean_price.index, mean_price.values, marker='o')

plt.title('Изменение средней стоимости квартир в Нагатинском Затоне по месяцам')
plt.xlabel('Месяц')
plt.ylabel('Средняя стоимость квартиры, руб.')

plt.xticks(mean_price.index, mean_price.index.strftime('%m.%Y'), rotation=90)

plt.grid(True)
plt.tight_layout()
plt.show()
````

2. Постройте гистограмму распределения цен на квартиры с общей площадью более 50 кв.метров

7. Постройте график изменения медианной цены на квартиру менее 50 кв. метров за период с 2011 до 2014 года*

\* Задание на доп. балл

````python
df['timestamp'] = pd.to_datetime(df['timestamp'])

df_small = df[df['full_sq'] < 50].copy()

df_small = df_small[
    (df_small['timestamp'].dt.year >= 2011) &
    (df_small['timestamp'].dt.year <= 2014)
]

median_price = df_small.groupby(
    pd.Grouper(key='timestamp', freq='ME')
)['price_doc'].median()

plt.figure(figsize=(14, 6))

plt.plot(
    median_price.index,
    median_price.values,
    marker='o',
    markersize=4
)

plt.title('Динамика медианной стоимости квартир площадью менее 50 кв. метров (2011–2014)')
plt.xlabel('Месяц')
plt.ylabel('Медианная стоимость квартиры, руб.')

plt.xticks(
    median_price.index,
    median_price.index.strftime('%m.%Y'),
    rotation=90
)

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````
